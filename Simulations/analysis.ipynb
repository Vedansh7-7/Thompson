```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import norm
```

```python
df= pd.read_csv('../outputs/visit_log.csv')
```

```python
df = df[df['arm_id'].isin([0, 3])]
df
```

```python
dfarm = df.groupby('arm_id')
dfarm
```

```python
df_dict = dfarm.count().to_dict()
```

```python
df_dict
```

```python
n0 = df_dict['day'][0]
n1 = df_dict['day'][3]
x0 = dfarm.sum().to_dict()['bought'][0]
x1 = dfarm.sum().to_dict()['bought'][3]
```

```python
from stats_tests import two_prop_z

two_prop_z(x0, n0, x1, n1)
```

```python
p1 = x1/n1
p0 = x0/n0
```

```python
# 95% confidence interval for p1 - p0
(p1 - p0) + 1.96 * np.sqrt(p0*(1-p0)/n1 + p1*(1-p1)/n0), (p1 - p0) - 1.96 * np.sqrt(p0*(1-p0)/n1 + p1*(1-p1)/n0)
```

```python
mean = (x1+x0)/(n1+n0)
print(x0,p0, x1,p1, mean)
```
